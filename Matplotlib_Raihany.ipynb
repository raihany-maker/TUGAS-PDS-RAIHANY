# 1. Scatter PLot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Import dataset
df = pd.read_csv("https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv")

# Prepare Data (Memilih beberapa kota sampel agar legenda rapi)
kota_pilihan = ['Kota Makassar', 'Kota Ambon', 'Kota Bandung', 'Kota Medan']
df_filter = df[df['Kota'].isin(kota_pilihan)]

categories = np.unique(df_filter['Kota'])
colors = [plt.cm.tab10(i/float(len(categories)-1)) for i in range(len(categories))]

# Draw Plot for Each Category
plt.figure(figsize=(12, 8), dpi=80, facecolor='w', edgecolor='k')

for i, category in enumerate(categories):
    sub_data = df_filter[df_filter.Kota == category]
    plt.scatter(sub_data['bm_ekonomi'], sub_data['PENGANGGURAN'],
                s=40, c=[colors[i]], label=str(category))

# Decorations
plt.gca().set(xlabel='Belanja Modal Ekonomi (bm_ekonomi)', ylabel='Tingkat Pengangguran')
plt.xticks(fontsize=12); plt.yticks(fontsize=12)
plt.title("Scatterplot of bm_ekonomi vs Pengangguran", fontsize=20)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# 2. Bubble Plot with Encircling

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.spatial import ConvexHull
import warnings; warnings.simplefilter('ignore')

sns.set_style("white")

# Step 1: Prepare Data
df = pd.read_csv("https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv")

kota_pilihan = ['Kota Makassar', 'Kota Ambon', 'Kota Bandung', 'Kota Medan']
df_filter = df[df['Kota'].isin(kota_pilihan)].dropna(subset=['bm_ekonomi', 'PENGANGGURAN'])

categories = np.unique(df_filter['Kota'])
colors = [plt.cm.tab10(i/float(len(categories)-1)) for i in range(len(categories))]

# Step 2: Draw Scatterplot with unique color for each category
fig = plt.figure(figsize=(12, 8), dpi=80, facecolor='w', edgecolor='k')

for i, category in enumerate(categories):
    sub_data = df_filter[df_filter.Kota == category]
    plt.scatter(sub_data['bm_ekonomi'], sub_data['PENGANGGURAN'],
                s=60, c=[colors[i]], label=str(category),
                edgecolors='black', linewidths=.5)

# Step 3: Encircling Function
def encircle(x, y, ax=None, **kw):
    if not ax: ax=plt.gca()
    p = np.c_[x, y]
    hull = ConvexHull(p)
    poly = plt.Polygon(p[hull.vertices,:], **kw)
    ax.add_patch(poly)

# Select data to be encircled (Contoh: Kota Makassar)
df_encircle_data = df_filter.loc[df_filter.Kota=='Kota Makassar', :]

# Draw polygon surrounding vertices
encircle(df_encircle_data.bm_ekonomi, df_encircle_data.PENGANGGURAN, ec="k", fc="gold", alpha=0.2)
encircle(df_encircle_data.bm_ekonomi, df_encircle_data.PENGANGGURAN, ec="firebrick", fc="none", linewidth=1.5)

# Step 4: Decorations
plt.gca().set(xlabel='BM Ekonomi', ylabel='Pengangguran')
plt.xticks(fontsize=12); plt.yticks(fontsize=12)
plt.title("Bubble Plot with Encircling (Kota Makassar)", fontsize=20)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# 3. Scatter Plot with Linear Regression Line of Best Fit

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Import Data
df = pd.read_csv("https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv")

# Filter data beberapa kota pilihan
df_select = df.loc[df.Kota.isin(['Kota Makassar', 'Kota Ambon', 'Kota Bandung', 'Kota Medan']), :]

# Plot
sns.set_style("white")

gridobj = sns.lmplot(x="bm_ekonomi", y="PENGANGGURAN", hue="Kota", data=df_select,
                     height=7, aspect=1.6, robust=True, palette='tab10',
                     scatter_kws=dict(s=60, linewidths=.7, edgecolors='black'))

# Decorations
plt.title("Scatterplot with Line of Best Fit Grouped by Kota", fontsize=18)
plt.xlabel("BM Ekonomi")
plt.ylabel("Pengangguran")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
