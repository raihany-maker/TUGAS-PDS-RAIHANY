# **Sel 1: Pengenalan Pandas Series & NumPy (#P1 - #P6)**

```python
import pandas as pd
import numpy as np

# #P1: Menampilkan array 1-dimensi
print("--- #P1: Pandas Series ---")
ds = pd.Series([2, 4, 6, 8, 10])
print(ds)
```

```python
# #P2: Series dari list & numpy dengan indeks kustom
print("\n--- #P2: Series & Indeks Kustom ---")
data = pd.Series(['a', 'b', 'c'])
np1 = np.array(['x', 'y', 'z'])
data_np = pd.Series(np1)
print("Data Series biasa:\n", data)
print("\nData Series dari NumPy:\n", data_np)

data_np = pd.Series(np1, index=[2, 3, 4])
print("\nData Series dengan indeks [2,3,4]:\n", data_np)
```

```python
# #P3: Operasi aritmatika pada Series
print("\n--- #P3: Operasi Aritmatika ---")
ds1 = pd.Series([2, 4, 6, 8, 10])
ds2 = pd.Series([1, 3, 5, 7, 9])
print("Penjumlahan (+):\n", ds1 + ds2)
print("\nPengurangan (-):\n", ds1 - ds2)
print("\nPerkalian (*):\n", ds1 * ds2)
print("\nPembagian (/):\n", ds1 / ds2)
```

```python
# #P4: Konversi Pandas Series ke Python List
print("\n--- #P4: Konversi Series ke List ---")
ds = pd.Series([2, 4, 6, 8, 10])
print("Pandas Series dan type:")
print(ds)
print(type(ds))

print("\nKonversi Pandas Series ke Python list:")
print(ds.tolist())
print(type(ds.tolist()))
```

```python
# #P5: Konversi NumPy array ke Pandas Series
print("\n--- #P5: Convert NumPy Array ke Series ---")
np_array = np.array([0, 20, 30, 40, 50])
print("NumPy array:", np_array)

new_np = pd.Series(np_array)
print("\nKonversi Pandas series:\n", new_np)
```

```python
# #P6: Generate angka random dengan RandomState
print("\n--- #P6: RandomState ---")
num_state = np.random.RandomState(100)
num_series = pd.Series(num_state.normal(10, 4, 20))
print(num_series)
```

# **Sel 2: Load Data (#P7)**

```python
# ==========================================
# Sel 2: Load Data Penjualan (#P7)
# ==========================================
import pandas as pd

# #P7.2: Load data langsung dari URL raw GitHub
url = 'https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/main/penjualan.csv'
df = pd.read_csv(url)

print("--- 4 Baris Pertama ---")
display(df.head(4))

print("--- 5 Baris Terakhir ---")
display(df.tail())
```

# **Sel 3: Ekstrak & Manipulasi DataFrame (#P8 - #P11)**

```python
# #P8: Menampilkan 1 Kolom (Pandas Series)
feb = df["feb"]
print("--- #P8: Kolom Feb (Series) ---")
print(feb.head())
```

```python
# #P9: Menampilkan Lebih dari Satu Kolom (DataFrame - Double Bracket)
print("\n--- #P9: Kolom Feb & Mar (DataFrame) ---")
display(df[['feb', 'mar']].head(3))
```

```python
# #P10: Tambah Kolom Teks Berisi Angka
penjual = ['penjual no ' + str(i) for i in range(1, 11)]
df['urutan'] = penjual
```

```python
# #P11: Ekstraksi Angka dari Teks dengan RegEx
df['no'] = df['urutan'].str.extract(r'(\d+)').astype('int64')

print("\n--- #P10 & #P11: DataFrame Setelah Edit Kolom ---")
display(df)
```

# **Sel 4: Memilih Data Berdasarkan Baris & Kolom (#P12)**

```python
# #P12.1: Menampilkan baris indeks berlabel 0
print("--- #P12.1: Baris Indeks 0 ---")
print("Via loc  :\n", df.loc[0])
print("\nVia iloc :\n", df.iloc[0])
```

```python
# #P12.2: Menampilkan baris 0 dan kolom 'jan'
print("\n--- #P12.2: Baris 0, Kolom 'jan' ---")
print("loc  :", df.loc[0, 'jan'])
print("iloc :", df.iloc[0, 1])
display(df[['jan']][:1])
```

```python
# #P12.3: Menampilkan subset baris & kolom tertentu
print("\n--- #P12.3: loc 0:2 vs iloc 0:3 ---")
display(df.loc[0:2, ['nama', 'jan', 'mar']])
display(df.iloc[0:3, [0, 1, 3]])
```

```python
# #P12.4: Menampilkan semua baris kolom nama & apr
print("\n--- #P12.4: Semua Baris Kolom nama & apr ---")
display(df.loc[:, ['nama', 'apr']])
```

```python
# #P12.5: Menampilkan semua baris kolom nama sampai feb
print("\n--- #P12.5: Kolom nama Sampai feb ---")
display(df.loc[:, 'nama':'feb'])
```

# **Sel 5: Memfilter Data (#P13 - #P14)**

```python
import re

# #P13.2: Filtering Berdasarkan Teks
print("--- Nama mengandung 'Andi' ---")
display(df[df['nama'].str.contains('Andi')])

print("--- Nama mengandung 'in' (Ignore Case) ---")
display(df[df['nama'].str.contains('in', flags=re.IGNORECASE)])

print("--- Feb bernilai 130 dan 220 ---")
feb_penjualan = [130, 220]
display(df[df['feb'].isin(feb_penjualan)])
```

```python
# #P13.1: Filtering Berdasarkan Angka (Satu Kondisi)
print("--- Jan == 200 ---")
display(df[df['jan'] == 200])

print("--- Jan > 200 ---")
display(df[df['jan'] > 200])
```

```python
# #P14: Filtering Menggunakan DataFrame Lain & Banyak Kondisi
d = {'tinggi': [320, 300], 'rendah': [95, 100]}
df1 = pd.DataFrame(data=d)

print("--- Jan yang ada di df1['tinggi'] ---")
display(df[df['jan'].isin(df1['tinggi'])])

print("--- Jan yang ada di df1['rendah'] ---")
display(df[df['jan'].isin(df1['rendah'])])

print("--- Jan Antara 100 dan 300 ---")
display(df[(df['jan'] > 100) & (df['jan'] < 300)])

print("--- Nama ada 'on' (Ignore Case) ATAU Jan > 200 ---")
display(df[(df['nama'].str.contains('on', flags=re.IGNORECASE)) | (df['jan'] > 200)])
```

# **Sel 6: Metode Pandas & Fungsi Apply (#P15)**

```python
# ==========================================
# Sel 6: Ringkasan Statistik & Apply (#P15)
# ==========================================

# Ringkasan statistik DataFrame
print("--- Ringkasan Statistik (df.describe) ---")
display(df.describe())

# #P15: Metode apply untuk fungsi kustom pada Series
print("\n--- #P15: Hasil Fungsi Apply ---")
data = [1, 1, 2, 3, 5, 8, 13, 21]
d_int = pd.Series(data)

def kecilbesar(num):
    if num < 3:
        return "kecil"
    elif num >= 3 and num < 13:
        return "sedang"
    else:
        return "besar"

hasil = d_int.apply(kecilbesar)
print(hasil)

# Informasi struktur DataFrame
print("\n--- Informasi DataFrame (df.info) ---")
df.info()
```
