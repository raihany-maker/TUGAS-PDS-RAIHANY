### **IMPORT DATASET DARI GITHUB**

```python
import pandas as pd

# load data CSV
url = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv"
df1 = pd.read_csv(url)

# ambil data penjualan bulan Januari (kolom 'jan')
data_januari = df1['jan']

# hitung
rata_rata = data_januari.mean()
varians = data_januari.var()
std_deviasi = data_januari.std()

# tampilkan hasil
print(f"Rata-rata Penjualan Januari      : {rata_rata}")
print(f"Varians Penjualan Januari        : {varians}")
print(f"Standar Deviasi Penjualan Januari: {std_deviasi}")
```
